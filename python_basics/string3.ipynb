```python
my_str= 'hello world'
upper_str= my_str.upper()
print(upper_str)
```

```python
my_str= 'hello WORLD'
lower_str= my_str.lower()
print(lower_str)
```

```python
my_str= '     hello world          '
trimmed_str= my_str.strip()
print(trimmed_str)
```

```python
my_str= 'hello world'
replaced_str= my_str.replace('hello', 'hi')
print(replaced_str)
```

```python
my_str= 'hello world'
split_str= my_str.split()
print(split_str)
```

```python
my_list= ['hello', 'world', 'python']
joined_str= ' '.join(my_list)
print(joined_str)
```

```python
my_str= 'hello world'
startwith_str= my_str.startswith('hello')
print(startwith_str)
```

```python
my_str= 'hello world'
endwith_str= my_str.endswith('world')
print(endwith_str)
```

```python
my_str= 'hello world'
o_count= my_str.count('o')
print(o_count)
```

```python
my_str= 'hello world'
capitalize_str= my_str.capitalize()
print(capitalize_str)
```

```python
my_str= 'hello world'
is_lower_str= my_str.islower()
print(is_lower_str)
```

```python
my_str= 'hello world'
title_str= my_str.title()
print(title_str)
```
