```python
#local scope
def my_func():
    my_var = 10
    print(my_var)
my_func()
```

```python
#enclosing sscope
def outer_func():
    msg = 'Hello'
    def inner_func():
        print(msg)
    inner_func()
outer_func()
```

```python
def outer_func():
    msg = 'Hello'
    res = ""
    def inner_func():
        nonlocal res
        res = "world"
        print(msg)
    inner_func()
    print(res)
outer_func()
```

```python
#global scope
var = 10
def change_var():
    global var
    var = 20
change_var()
print(var)
```
